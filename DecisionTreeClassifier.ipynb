```python
from sklearn.datasets import load_breast_cancer
```

```python
from sklearn.model_selection import train_test_split
data = load_breast_cancer()
X, y = data.data, data.target
```

```python
y[0:25]
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2,  random_state=3)
```

```python
tree = DecisionTreeClassifier(random_state=0)
tree.fit(X_train, y_train)
```

```python
print("Test Accuracy:", tree.score(X_test, y_test))
print("Train Accuracy:", tree.score(X_train, y_train))
```

```python
y_pred = tree.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(15,10))

plot_tree(tree, feature_names = data.feature_names, class_names = data.target_names , filled=True, fontsize = 8    )
plt.show()
```

```python
import numpy as np

def gini(y):
  p = np.bincount(y)/len(y)
  return 1 - np.sum(p**2)

print("gini of a node:"), gini(y_train)
```

```python
tree2 = DecisionTreeClassifier(random_state=0, max_depth = 4)
tree2.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(15,10))

plot_tree(tree2, feature_names = data.feature_names, class_names = data.target_names , filled=True, fontsize = 8    )
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt

depths = range(1, 10)
train_scores, test_scores = [], []

for d in depths:
    t = DecisionTreeClassifier(max_depth=d, random_state=42)
    t.fit(X_train, y_train)

    train_scores.append(t.score(X_train, y_train))
    test_scores.append(t.score(X_test, y_test))

plt.plot(depths, train_scores, marker="o", label="Train")
plt.plot(depths, test_scores, marker="o", label="Test")

plt.xlabel("max_depth")
plt.ylabel("Accuracy")
plt.title("Decision Tree Accuracy vs. Depth")
plt.xticks(list(depths))
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
