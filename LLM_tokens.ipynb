```python
import tiktoken
```

```python
encoding = tiktoken.get_encoding("cl100k_base")
```

```python
text = "Language models predict the next token."
```

```python
tokens = encoding.encode(text)
```

```python
for token_id in tokens:
    piece = encoding.decode([token_id])
    print(f"  {token_id:>6} -> {piece!r}")
```
